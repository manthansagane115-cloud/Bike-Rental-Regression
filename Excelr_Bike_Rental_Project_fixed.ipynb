```python

# FEATURE ENGINEERING


# Extract useful information from the date
df['day'] = df['dteday'].dt.day
df['weekofyear'] = df['dteday'].dt.isocalendar().week.astype(int)

# Weekend indicator
df['is_weekend'] = df['weekday'].isin([0, 6]).astype(int)

df.head()
```

```python
# Create a rush hour feature

df['rush_hour'] = df['hr'].isin([
    7, 8, 9, 17, 18, 19
]).astype(int)

df[['hr', 'rush_hour']].head(20)
```

```python
# Create time-of-day categories

def time_of_day(hour):
    if hour < 6:
        return 'Night'
    elif hour < 12:
        return 'Morning'
    elif hour < 17:
        return 'Afternoon'
    elif hour < 21:
        return 'Evening'
    else:
        return 'Night'

df['time_of_day'] = df['hr'].apply(time_of_day)

df[['hr', 'time_of_day']].head(20)

#This converts the 24-hour hr variable into meaningful business categories.
```

```python
# Cyclical encoding for hour

#Hours are cyclical: hour 23 and hour 0 are actually close to each other. 
#A normal numerical representation does not capture this properly.

df['hr_sin'] = np.sin(2 * np.pi * df['hr'] / 24)
df['hr_cos'] = np.cos(2 * np.pi * df['hr'] / 24)

df[['hr', 'hr_sin', 'hr_cos']].head(10)
```

```python
# Cyclical encoding for month

df['mnth'] = pd.to_numeric(df['mnth'], errors='coerce')

df['month_sin'] = np.sin(2 * np.pi * df['mnth'] / 12)
df['month_cos'] = np.cos(2 * np.pi * df['mnth'] / 12)

df[['mnth', 'month_sin', 'month_cos']].head(10)
```

```python
# Cyclical encoding for weekday

df['weekday_sin'] = np.sin(2 * np.pi * df['weekday'] / 7)
df['weekday_cos'] = np.cos(2 * np.pi * df['weekday'] / 7)

df[['weekday', 'weekday_sin', 'weekday_cos']].head(10)
```

```python
# Temperature difference between actual and feeling temperature

df['temp_difference'] = df['atemp'] - df['temp']

df[['temp', 'atemp', 'temp_difference']].head()
```

```python
# Weather comfort feature

df['comfort_index'] = df['temp'] * (1 - df['hum'])

df[['temp', 'hum', 'comfort_index']].head()
```

```python
df.head()
```

```python

# DEFINE FEATURES AND TARGET

# Target variable
y = df['cnt_log']

# Remove target and leakage columns
X = df.drop(
    columns=[
        'cnt',
        'cnt_log',
        'casual',
        'registered'
    ]
)

print("Feature shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# Remove identifier column

if 'instant' in X.columns:
    X = X.drop(columns=['instant'])

print("Feature shape after removing ID:", X.shape)
```

```python

# IDENTIFY NUMERICAL AND CATEGORICAL FEATURES


categorical_features = X.select_dtypes(
    include=['object']
).columns.tolist()

numerical_features = X.select_dtypes(
    include=['int64', 'float64', 'int32', 'float32']
).columns.tolist()

print("Categorical Features:")
print(categorical_features)

print("\nNumerical Features:")
print(numerical_features)
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    handle_unknown='ignore',
    sparse_output=False
)

X_encoded_cat = encoder.fit_transform(
    X[categorical_features]
)

X_encoded_cat = pd.DataFrame(
    X_encoded_cat,
    columns=encoder.get_feature_names_out(categorical_features),
    index=X.index
)

X_encoded_cat.head()
```

```python
# Combine numerical and encoded categorical features

X_numerical = X[numerical_features].reset_index(drop=True)
X_encoded_cat = X_encoded_cat.reset_index(drop=True)

X_final = pd.concat(
    [X_numerical, X_encoded_cat],
    axis=1
)

print("Final Feature Matrix Shape:", X_final.shape)

X_final.head()
```

```python
# Check missing values in final feature matrix

print("Missing values:", X_final.isnull().sum().sum())
```

```python

# FINAL FEATURE ENGINEERING SUMMARY


print("Original Dataset Shape :", df.shape)
print("Final Feature Shape    :", X_final.shape)
print("Target Shape           :", y.shape)

print("\nTarget Variable:")
print("cnt_log")

print("\nFinal Features:")
print(X_final.columns.tolist())
```

# 4.MODEL BUILDING

```python
# Import libraries required for model building and evaluation
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
# Split the dataset into training and testing sets 
X_train, X_test, y_train, y_test = train_test_split(
    X_final, y, test_size=0.2, random_state=42
)

print("Training set shape:", X_train.shape, y_train.shape)
print("Testing set shape :", X_test.shape, y_test.shape)
```

```python

# Define the candidate models
models = {
    "Decision Tree": DecisionTreeRegressor(max_depth=10, random_state=42),
    "Random Forest": RandomForestRegressor(n_estimators=200, max_depth=15, random_state=42, n_jobs=-1),
    "Gradient Boosting": GradientBoostingRegressor(n_estimators=200, learning_rate=0.1, max_depth=4, random_state=42)
}

results = []
predictions = {}

for name, model in models.items():
    # Train the model
    model.fit(X_train, y_train)

    # Predict on log scale, then convert back to actual rental counts
    y_pred_log = model.predict(X_test)
    y_pred = np.expm1(y_pred_log)
    y_actual = np.expm1(y_test)

    # Evaluate performance
    rmse = np.sqrt(mean_squared_error(y_actual, y_pred))
    mae = mean_absolute_error(y_actual, y_pred)
    r2 = r2_score(y_actual, y_pred)

    results.append({"Model": name, "RMSE": rmse, "MAE": mae, "R2 Score": r2})
    predictions[name] = y_pred

    print(f"\n{name}")
    print(f"RMSE : {rmse:.4f}")
    print(f"MAE  : {mae:.4f}")
    print(f"R2   : {r2:.4f}")
```

```python

# Consolidate results into a comparison table, sorted by R2 Score
results_df = pd.DataFrame(results).sort_values(by="R2 Score", ascending=False).reset_index(drop=True)
results_df
```

```python

# Bar plots comparing RMSE, MAE, and R2 Score across models
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
metrics = ['RMSE', 'MAE', 'R2 Score']
colors = ['#4C72B0', '#DD8452', '#55A868']

for ax, metric in zip(axes, metrics):
    sns.barplot(data=results_df, x='Model', y=metric, hue='Model', legend=False, palette=colors, ax=ax)
    ax.set_title(f'Model Comparison - {metric}')
    ax.set_xlabel('')
    ax.tick_params(axis='x', rotation=15)

plt.tight_layout()
plt.show()
```

```python

# Identify the most influential features according to the Random Forest model
rf_model = models['Random Forest']
importances = pd.Series(rf_model.feature_importances_, index=X_train.columns)
top_features = importances.sort_values(ascending=False).head(15)

plt.figure(figsize=(9,7))
sns.barplot(x=top_features.values, y=top_features.index, hue=top_features.index, legend=False, palette='viridis')
plt.title('Top 15 Feature Importances - Random Forest')
plt.xlabel('Importance')
plt.ylabel('Feature')
plt.tight_layout()
plt.show()
```

```python

# Scatter plot of actual vs predicted rentals for the best-performing model
y_actual = np.expm1(y_test)
best_pred = predictions['Random Forest']

plt.figure(figsize=(7,7))
plt.scatter(y_actual, best_pred, alpha=0.3, color='#4C72B0')
plt.plot([y_actual.min(), y_actual.max()], [y_actual.min(), y_actual.max()], 'r--', lw=2)
plt.xlabel('Actual Rental Count')
plt.ylabel('Predicted Rental Count')
plt.title('Random Forest: Actual vs Predicted Bike Rentals')
plt.tight_layout()
plt.show()
```


**Model Building Summary:**

Three regression algorithms were trained to predict bike rental demand using an 80/20 train-test split (13,903 training rows and 3,476 test rows). The target variable (log-transformed rental count) was modeled and predictions were converted back to actual rental counts for evaluation.

- **Random Forest** was the best-performing model, achieving the lowest RMSE (**41.30**), lowest MAE (**24.45**), and highest R² (**0.9461**), meaning it explains about 94.6% of the variance in bike rental demand.
- **Gradient Boosting** performed nearly as well (R² = 0.9458, RMSE = 41.42), showing that boosting can match bagging-based ensembles on this dataset with the right hyperparameters.
- **Decision Tree** (a single, non-ensembled model) performed noticeably worse (R² = 0.8750, RMSE = 62.92), confirming that ensemble methods reduce variance and generalize better than a single tree.
- The feature importance analysis shows that **hour of day (`hr`)** is by far the strongest predictor of rental demand, followed by its cyclical encodings (`hr_sin`, `hr_cos`), temperature, and comfort index — consistent with the hourly demand patterns observed during EDA.

**Conclusion:** Random Forest is recommended as the final model for bike rental demand prediction, with Gradient Boosting as a strong alternative if further hyperparameter tuning is desired.

# Hyperparameter Tuning

```python
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
```

```python
#Decision Tree Tuning
dt_param_grid = {
    'max_depth': [10, 15, 20],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

dt_grid = GridSearchCV(
    DecisionTreeRegressor(random_state=42),
    dt_param_grid,
    cv=5,
    scoring='neg_mean_squared_error',
    n_jobs=-1
)

dt_grid.fit(X_train, y_train)

print("Best Decision Tree Parameters:")
print(dt_grid.best_params_)
```

```python
#Random Forest Tuning
rf_param_grid = {
    'n_estimators': [50, 100],
    'max_depth': [10, 15],
    'min_samples_split': [2],
    'min_samples_leaf': [1]
}

rf_grid = GridSearchCV(
    RandomForestRegressor(random_state=42, n_jobs=-1),
    rf_param_grid,
    cv=3,
    scoring='neg_mean_squared_error',
    n_jobs=-1
)

rf_grid.fit(X_train, y_train)

print("Best Random Forest Parameters:")
print(rf_grid.best_params_)
```

```python
#Gradient Boosting Tuning
gb_param_grid = {
    'n_estimators': [50, 100],
    'learning_rate': [0.1],
    'max_depth': [2, 3]
}

gb_grid = GridSearchCV(
    GradientBoostingRegressor(random_state=42),
    gb_param_grid,
    cv=2,
    scoring='neg_mean_squared_error',
    n_jobs=-1
)

gb_grid.fit(X_train, y_train)

print("Best Gradient Boosting Parameters:")
print(gb_grid.best_params_)
```

```python
#Best Parameters
tuning_summary = pd.DataFrame({
    "Model": [
        "Decision Tree",
        "Random Forest",
        "Gradient Boosting"
    ],
    "Best Parameters": [
        dt_grid.best_params_,
        rf_grid.best_params_,
        gb_grid.best_params_
    ]
})

tuning_summary
```

Hyperparameter Tuning Summary:

Hyperparameter tuning was performed for the three regression algorithms 
— Decision Tree, Random Forest, and Gradient Boosting — using GridSearchCV with cross-validation. Different combinations of model parameters were tested to identify the best settings for each model.

1) Decision Tree: The best parameters obtained were max_depth = 15, min_samples_leaf = 4, and min_samples_split = 10.
2) Random Forest: Hyperparameters such as n_estimators, max_depth, and min_samples_leaf were tuned, and the best parameter combination was identified using cross-validation.
3) Gradient Boosting: Parameters including learning_rate, max_depth, and n_estimators were tuned, and the best combination was selected.

GridSearchCV helped automatically test different hyperparameter combinations and select the best settings based on cross-validation performance.
The tuned models were then passed to the Model Evaluation stage for comparison using MAE, MSE, RMSE, and R².

Conclusion: Hyperparameter tuning successfully identified optimized parameter combinations for all three regression models, providing tuned models for the next Model Evaluation stage.

```python
# Test the prediction function

sample_input = {
    "season": "summer",
    "yr": 2012,
    "mnth": 6,
    "hr": 17,
    "holiday": "No",
    "weekday": 2,
    "workingday": "Working Day",
    "weathersit": "Clear",
    "temp": 0.60,
    "atemp": 0.60,
    "hum": 0.50,
    "windspeed": 0.20,
    "dteday": "2012-06-15"
}

def predict_rentals(input_data):
    row = input_data.copy()

    # IMPORTANT: the OneHotEncoder was fit on the ORIGINAL text labels in the
    # dataset, so the categorical columns must stay as those exact strings
    # (not be converted to integers) before being passed to encoder.transform().
    season_map = {
        "spring": "springer",
        "springer": "springer",
        "summer": "summer",
        "fall": "fall",
        "autumn": "fall",
        "winter": "winter"
    }
    weather_map = {
        "clear": "Clear",
        "mist": "Mist",
        "light snow": "Light Snow",
        "heavy rain": "Heavy Rain"
    }
    holiday_map = {
        "no": "No",
        "yes": "Yes"
    }
    workingday_map = {
        "working day": "Working Day",
        "yes": "Working Day",
        "no work": "No work",
        "no": "No work"
    }

    row["season"] = season_map.get(str(row["season"]).lower(), row["season"])
    row["weathersit"] = weather_map.get(str(row["weathersit"]).lower(), row["weathersit"])
    row["holiday"] = holiday_map.get(str(row["holiday"]).lower(), row["holiday"])
    row["workingday"] = workingday_map.get(str(row["workingday"]).lower(), row["workingday"])

    # "yr" is a NUMERICAL feature in the training data and holds the literal
    # year (2011.0 / 2012.0) -- it must NOT be collapsed into a 0/1 flag.
    row["yr"] = float(row["yr"])

    row["dteday"] = pd.to_datetime(row["dteday"])

    # Recreate engineered features
    row["day"] = row["dteday"].day
    row["weekofyear"] = row["dteday"].isocalendar().week
    row["is_weekend"] = int(row["weekday"] in [0, 6])
    row["rush_hour"] = int(row["hr"] in [7, 8, 9, 17, 18, 19])

    row["time_of_day"] = time_of_day(row["hr"])
    row["hr_sin"] = np.sin(2 * np.pi * row["hr"] / 24)
    row["hr_cos"] = np.cos(2 * np.pi * row["hr"] / 24)
    row["month_sin"] = np.sin(2 * np.pi * row["mnth"] / 12)
    row["month_cos"] = np.cos(2 * np.pi * row["mnth"] / 12)
    row["weekday_sin"] = np.sin(2 * np.pi * row["weekday"] / 7)
    row["weekday_cos"] = np.cos(2 * np.pi * row["weekday"] / 7)
    row["temp_difference"] = row["atemp"] - row["temp"]
    row["comfort_index"] = row["temp"] * (1 - row["hum"])

    input_df = pd.DataFrame([row])

    numeric_part = input_df[numerical_features].reset_index(drop=True)
    categorical_part = pd.DataFrame(
        encoder.transform(input_df[categorical_features]),
        columns=encoder.get_feature_names_out(categorical_features)
    )

    final_input = pd.concat([numeric_part, categorical_part], axis=1)
    final_input = final_input.reindex(columns=feature_columns, fill_value=0)

    prediction_log = rf_grid.best_estimator_.predict(final_input)[0]
    return float(np.expm1(prediction_log))


prediction = predict_rentals(sample_input)

print("Predicted Bike Rentals:", round(prediction))
```
